```python
df["city"] = df["city"].fillna("Unknown")

df["player_of_match"] = df["player_of_match"].fillna("Unknown")

df["winner"] = df["winner"].fillna("Unknown")
```

```python
print(df[df["id"].isnull()])

print(df["method"].value_counts(dropna=False))

print(df[["result", "result_margin", "target_runs", "target_overs", "method"]].head(20))
```

```python
df[df["winner"].isnull()][["team1", "team2", "result", "winner"]]
df[df["result_margin"].isnull()][["team1", "team2", "winner", "result", "result_margin"]]
df[df["target_runs"].isnull()][["team1", "team2", "result", "winner", "target_runs", "target_overs"]]
df[df["city"].isnull()][["team1", "team2", "venue", "city"]]
```

```python
print(df[df["city"].isnull()][["team1", "team2", "venue"]].head(10))
print(df[df["winner"].isnull()][["team1", "team2", "result", "winner"]])
```

```python
print(df.isnull().sum())
```

```python
df["method"] = df["method"].fillna("No D/L")
print(df[df["result_margin"].isnull()][
    ["team1", "team2", "winner", "result", "result_margin"]
])
print(df[df["target_runs"].isnull()][
    ["team1", "team2", "winner", "result", "target_runs", "target_overs"]
])
print(df[df["id"].isnull()])
```

```python
df.drop(columns=["id"], inplace=True)
df.isnull().sum()
```

```python
print(df.duplicated().sum())
df.info()
df.describe()
```

```python
df["date"] = pd.to_datetime(df["date"])


wins = df["winner"].value_counts()

print(wins)
```

```python
import matplotlib.pyplot as plt

wins = df["winner"].value_counts()

plt.bar(wins.index, wins.values)

plt.xlabel("Teams")
plt.ylabel("Number of Wins")
plt.title("Number of Wins by Team")

plt.xticks(rotation=90)
plt.show()
```
