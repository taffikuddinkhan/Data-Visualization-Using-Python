```python
import matplotlib.pyplot as plt

month=[1,2,3,4,5]
companya=[30,20,15,40,50]
companyb=[10,50,30,60,20]

plt.xlabel("person buy")
plt.ylabel("toy sold")
plt.title("items sold chart",color="r")
plt.grid()
# plt.xlim(1,10) # control the range of x axis
# plt.ylim(0,50)
plt.plot(
month,companya,
color="blue",
linestyle="-",
linewidth="2",
marker="o",
markersize=10,
markerfacecolor="yellow",
markeredgecolor="black",
alpha=0.5, # maintain transperence of the line
markevery=1 # maintain marker points 
)

plt.plot(
month,companyb,
color="green",
linestyle="-",
linewidth="2",
marker="o",
markersize=10,
markerfacecolor="yellow",
markeredgecolor="black",
alpha=0.5, # maintain transperence of the line
markevery=1 # maintain marker points 
)
plt.text(1,3,"average of companty b")  # used to add text in a chart in anyplace
plt.show()
```

```python
laptops = ["asus","lenovo","dell","infinix"]
sales = [100,30,40,500]
plt.bar(
laptops,sales,
color="red",
edgecolor="black",
label="this month sales",
hatch="//", # add pattern inside the bins
  
)
plt.legend()
plt.show()

# horizontal barchart

laptops = ["asus","lenovo","dell","infinix"]
sales = [100,30,40,500]
plt.barh(
laptops,sales,
color="green",
edgecolor="black",
label="this month sales",
hatch="//", # add pattern inside the bins
  
)
plt.legend()
plt.show()


#multiple charts at a time 

laptops = ["asus","lenovo","dell","infinix"]
mobiles = ["vivo","oppo","samsung","realme"]
sales = [100,30,40,500]
plt.bar(
laptops,sales,
color="green",
edgecolor="black",
label="this month sales",
hatch="//", # add pattern inside the bins
)
plt.xlim(0,10)

plt.bar(
mobiles,sales,
color="blue",
edgecolor="black",
label="this month sales",
hatch="//", # add pattern inside the bins
)
plt.tight_layout()
plt.legend()
plt.show()
```
