```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5, 6]
y = [10, 25, 20, 35, 30, 45]

plt.plot(
    x,
    y,
    color="blue",
    marker="o",
    markersize=8,
    linewidth=2.5,
    linestyle="-",
    alpha=0.8
)

plt.title("Line Plot", fontsize=18)
plt.xlabel("X Values", fontsize=14)
plt.ylabel("Y Values", fontsize=14)

plt.xticks(fontsize=11)
plt.yticks(fontsize=11)

plt.grid(
    True,
    linestyle="--",
    alpha=0.5
)

plt.show()
```

```python
import matplotlib.pyplot as plt

subjects = ["Python", "Java", "SQL", "HTML", "CSS"]
marks = [85, 90, 78, 88, 82]

plt.bar(
    subjects,
    marks,
    color="green",
    width=0.6,
    alpha=0.8,
    edgecolor="black",
    linewidth=1.5
)

plt.title("Student Marks", fontsize=18)
plt.xlabel("Subjects", fontsize=14)
plt.ylabel("Marks", fontsize=14)

plt.xticks(fontsize=11)
plt.yticks(fontsize=11)

plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.5
)

plt.show()
```

```python
import matplotlib.pyplot as plt

x = [10, 20, 30, 40, 50, 60]
y = [15, 25, 20, 35, 40, 50]

plt.scatter(
    x,
    y,
    color="red",
    marker="*",
    s=150,
    alpha=0.7,
    edgecolor="black",
    linewidth=1.5
)

plt.title("Scatter Plot", fontsize=18)
plt.xlabel("X Values", fontsize=14)
plt.ylabel("Y Values", fontsize=14)

plt.xticks(fontsize=11)
plt.yticks(fontsize=11)

plt.grid(
    True,
    linestyle="--",
    alpha=0.5
)

plt.show()
```
