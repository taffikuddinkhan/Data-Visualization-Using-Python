```python
import matplotlib.pyplot as plt

day = [10, 20, 30, 40, 50, 60]  #day is variable and value are parameters
month = [1, 2, 3, 4, 5, 6]
color = [10, 20, 30, 40, 50, 60]

plt.plot(day, month ,alpha=0.5, c=color, marker="*",cmap="magma")
plt.grid()
plt.colorbar()
plt.title("linechart", color="r", fontsize=30)
plt.xlabel("day", color="g", fontsize=10)
plt.ylabel("month", color="y", fontsize=20)
plt.show()
```

```python
import matplotlib.pyplot as plt

day = [10, 20, 30, 40, 50, 60]       # X-axis values
month = [1, 2, 3, 4, 5, 6]           # Y-axis values

plt.plot(
    day,
    month,
    color="blue",
    alpha=0.7,
    marker="*",
    linewidth=2
)

plt.grid()

plt.title("Line Chart", colo="red", fontsize=30)
plt.xlabel("Day", color="green", fontsize=10)
plt.ylabel("Month", color="orange", fontsize=20)

plt.show()
```
